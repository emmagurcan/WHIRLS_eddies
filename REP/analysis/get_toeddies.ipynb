```python
from pathlib import Path
import requests

# ------------------------------------------------------------------
# Configuration
# ------------------------------------------------------------------

BASE_URL = "https://thredds-x.ipsl.fr/thredds/fileServer/WHIRLS/PRODUCTS/TOEDDIES"

SAVE_ROOT = Path("/Volumes/science/MD252_WHIRLS/REP/data/Eddies")

FILE_TYPES = [
    "anticyclones_centers",
    "anticyclones_maxspeed",
    "anticyclones_outcontour",
    "cyclones_centers",
    "cyclones_maxspeed",
    "cyclones_outcontour",
]


# ------------------------------------------------------------------
# Download function
# ------------------------------------------------------------------

def download_toeddies(day):
    """
    Download the six TOEDDIES geojson files for a given day.

    Parameters
    ----------
    day : str
        Date in the format YYYY-MM-DD.
    """

    out_dir = SAVE_ROOT / day
    out_dir.mkdir(parents=True, exist_ok=True)

    existing_files = list(out_dir.glob("*.geojson"))

    if len(existing_files) >= 6:
        print(f"{day}: already contains 6 files. Skipping.")
        return

    print(f"{day}: downloading missing files...")

    for file_type in FILE_TYPES:

        filename = f"{day}_{file_type}.geojson"
        outfile = out_dir / filename

        if outfile.exists():
            print(f"  ✓ {filename} already exists")
            continue

        url = f"{BASE_URL}/{filename}"

        print(f"  Downloading {filename}")

        r = requests.get(url, stream=True)

        if r.status_code == 200:
            with open(outfile, "wb") as f:
                for chunk in r.iter_content(chunk_size=8192):
                    if chunk:
                        f.write(chunk)
            print("     done")
        else:
            print(f"     ERROR ({r.status_code})")


# ------------------------------------------------------------------
# Example
# ------------------------------------------------------------------

if __name__ == "__main__":

    day = "2026-07-16"

    download_toeddies(day)
```
